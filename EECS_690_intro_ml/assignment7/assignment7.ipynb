```python
import pandas as pd
import numpy as np
import random
```

```python
print("Part 1: Policy Iteration RL Algorithm")
gamma = 1
rewardSize = -1
gridSize = 5
terminationStates = [[0,0], [gridSize-1, gridSize-1]]
actions = [[-1, 0], [1, 0], [0, 1], [0, -1]]
numIterations = 1000

def actionRewardFunction(initialPosition, action):
    
    if initialPosition in terminationStates:
        return initialPosition, 0
    
    reward = rewardSize
    finalPosition = np.array(initialPosition) + np.array(action)
    if -1 in finalPosition or gridSize in finalPosition: 
        finalPosition = initialPosition
        
    return finalPosition, reward

```

```python
# Initialization
valueMap = np.zeros((gridSize, gridSize))
states = [[i, j] for i in range(gridSize) for j in range(gridSize)]
# values of the value function at step 0
valueMap
```

```python
# policy evaluation
deltas = []
sums = []
sum_pre = 0
thld = 0.05
for it in range(numIterations):
    copyValueMap = np.copy(valueMap)
    deltaState = []
    for state in states:
        weightedRewards = 0
        for action in actions:
            finalPosition, reward = actionRewardFunction(state, action)
            weightedRewards += (1/len(actions))*(reward+(gamma*valueMap[finalPosition[0], finalPosition[1]]))
        deltaState.append(np.abs(copyValueMap[state[0], state[1]]-weightedRewards))
        copyValueMap[state[0], state[1]] = weightedRewards
    deltas.append(deltaState)
    _sum = np.sum(copyValueMap)
    sums.append(_sum)
    valueMap = copyValueMap
    if it in [0, 9, numIterations - 1]:
        print("Iteration {}".format(it + 1))
        print(f"sum:{_sum}")
        print(valueMap)
        print("")
    # stop if sum diff is less than a pre-set threshold
    if abs((_sum - sum_pre) / sum_pre) < thld:
        print(f"Iteration {it + 1}. Sum delta smaller than threshold({thld}), stop.")
        print(f"sum:{_sum}")
        print(valueMap)
        print("")
        break
    sum_pre = _sum
```

```python
print("Part 2: Value Iteration RL Algorithm")

gamma = 1
rewardSize = -1
gridSize = 5
terminationStates = [[0,0], [gridSize-1, gridSize-1]]
actions = [[-1, 0], [1, 0], [0, 1], [0, -1]]
numIterations = 1000

# Initialization
valueMap = np.zeros((gridSize, gridSize))
states = [[i, j] for i in range(gridSize) for j in range(gridSize)]
# values of the value function at step 0

# value evaluation
deltas = []
sums = []
sum_pre = 0
thld = 0.05
for it in range(numIterations):
    copyValueMap = np.copy(valueMap)
    deltaState = []
    for state in states:
        weightedRewards = []
        for action in actions:
            finalPosition, reward = actionRewardFunction(state, action)
            #print(f"{reward}+{gamma}*{valueMap[finalPosition[0], finalPosition[1]]}")
            #_reward = (1/len(actions)) * (reward+(gamma*valueMap[finalPosition[0], finalPosition[1]]))
            _reward = reward + gamma * valueMap[finalPosition[0], finalPosition[1]]
            weightedRewards.append(_reward)
        best_reward = max(weightedRewards)
        #print(f"best_reward:{best_reward}")
        deltaState.append(np.abs(copyValueMap[state[0], state[1]] - best_reward))
        copyValueMap[state[0], state[1]] = best_reward
    deltas.append(deltaState)
    _sum = np.sum(copyValueMap)
    sums.append(_sum)
    valueMap = copyValueMap
    if it in [0, 1, 2, 9, numIterations - 1]:
        print("Iteration {}".format(it + 1))
        print(f"sum:{_sum}")
        print(valueMap)
        print("")
    # stop if sum diff is less than a pre-set threshold
    if abs((_sum - sum_pre) / sum_pre) < thld:
        print(f"Iteration {it + 1}. Sum delta smaller than threshold({thld}), stop.")
        print(f"sum:{_sum}")
        print(valueMap)
        print("")
        break
    sum_pre = _sum
```
